# Fruit Sales Analysis

This notebook demonstrates a simple reproducible data-analysis project.
It uses a small built-in dataset of monthly fruit sales and creates summary tables and charts.

**Project goal:** compare sales of three fruits and identify monthly sales patterns.

```python
import pandas as pd
import matplotlib.pyplot as plt

print('Libraries loaded successfully.')
```

```python
# A small dataset is included directly in the notebook, so the project is reproducible.
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
    'Apple': [120, 135, 128, 145, 160, 172, 180, 176, 168, 155, 148, 162],
    'Banana': [100, 110, 118, 125, 130, 142, 150, 158, 152, 145, 138, 146],
    'Orange': [90, 96, 105, 115, 122, 135, 142, 140, 132, 125, 118, 130]
}

df = pd.DataFrame(data)
df
```

```python
# Check the structure of the dataset.
print('Rows:', len(df))
print('Columns:', list(df.columns))
print('\nMissing values:')
print(df.isna().sum())
```

```python
# Calculate total and average sales for each fruit.
summary = pd.DataFrame({
    'Total Sales': df[['Apple', 'Banana', 'Orange']].sum(),
    'Average Monthly Sales': df[['Apple', 'Banana', 'Orange']].mean().round(2),
    'Maximum Monthly Sales': df[['Apple', 'Banana', 'Orange']].max()
})

summary
```

```python
# Find the month with the highest combined sales.
df['Total'] = df[['Apple', 'Banana', 'Orange']].sum(axis=1)
best_month = df.loc[df['Total'].idxmax(), 'Month']
best_value = df['Total'].max()

print(f'Highest combined sales occurred in {best_month}: {best_value} units.')
```

```python
# Visualize monthly sales trends.
plt.figure(figsize=(10, 5))
plt.plot(df['Month'], df['Apple'], marker='o', label='Apple')
plt.plot(df['Month'], df['Banana'], marker='o', label='Banana')
plt.plot(df['Month'], df['Orange'], marker='o', label='Orange')
plt.title('Monthly Fruit Sales')
plt.xlabel('Month')
plt.ylabel('Sales (units)')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Compare total sales.
totals = df[['Apple', 'Banana', 'Orange']].sum().sort_values(ascending=False)

plt.figure(figsize=(7, 4))
totals.plot(kind='bar')
plt.title('Total Sales by Fruit')
plt.xlabel('Fruit')
plt.ylabel('Total Sales (units)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print('Ranking by total sales:')
for fruit, value in totals.items():
    print(f'{fruit}: {value} units')
```
